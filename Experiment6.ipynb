```python
#5. Check whether a year is a leap year.

year = int(input("Enter a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
     print(year, "is a leap year")
else:
     print(year, "is not a leap year")
```

```python
#6.
date_str = input("Enter date (YYYY-MM-DD): ")

parts = date_str.split('-')
if len(parts) == 3 and all(p.isdigit() for p in parts):
    year = int(parts[0])
    month = int(parts[1])
    day = int(parts[2])

    print(f"Original: {year}-{month:02d}-{day:02d}")

    # Manual increment logic
    day += 1

    # Month days array (non-leap year base)
    days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

    if day > days_in_month[month-1]:
        day = 1
        month += 1
        if month > 12:
            month = 1
            year += 1

    print(f"Next day: {year}-{month:02d}-{day:02d}")
else:
    print("Invalid format! Use YYYY-MM-DD (e.g., 2026-02-09)")
```

```python
''' Problem Statement :
Write a python program to calculate the gross salary of an employee'''

basic = float(input("Enter Basic Salary: "))
if basic <= 10000:
    hra = basic * 0.20
    da = basic * 0.8
elif  basic <= 20000:
   hra = basic * 0.25
   da = basic * 0.9
else :
   hra = basic * 0.30
   da = basic * 0.95

gross = basic + hra + da
print("Gross Salary = ", gross)
```
